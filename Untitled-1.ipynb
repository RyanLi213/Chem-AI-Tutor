### General **Validation** function for testing

```python
def validate_molecule(smile_string: str):
    if isinstance(smile_string, str):
        smile_string = [smile_string]

    results = []
    for smi in smile_string:
        mol = Chem.MolFromSmiles(smi)
        if mol is None:
            results.append({"Valid":False, "Stereocenters": []})
            continue

        # converts molecule to RDKit canon SMILES string and re-parse canon string back to molecule so atoms are numbered in canonical order
        canon_smiles = Chem.MolToSmiles(mol)
        mol = Chem.MolFromSmiles(canon_smiles)

        rdCIPLabeler.AssignCIPLabels(mol)

        stereocenters = []
        for atom in mol.GetAtoms():
            if atom.HasProp("_CIPCode"):
                stereocenters.append({
                    "Atom index": atom.GetIdx(),
                    "label": atom.GetProp("_CIPCode")
                })
        results.append({"Valid": True, 
               "Number of stereocenters": len(stereocenters), 
               "Stereocenters": stereocenters })

    return results
```

### Decimer Pipeline function

```python
from rdkit import Chem
from rdkit.Chem import Draw
from DECIMER import predict_SMILES

# Step 1: generate a test image of a known molecule (aspirin)
original_smiles = "CC(=O)Oc1ccccc1C(=O)O"
mol = Chem.MolFromSmiles(original_smiles)
Draw.MolToFile(mol, "aspirin_test.png", size=(400, 400))

# Step 2: feed it through DECIMER
predicted_smiles = predict_SMILES("aspirin_test.png")
print("Original: ", original_smiles)
print("Predicted:", predicted_smiles)

# Step 3: compare using RDKit (SMILES strings can differ in formatting
# even when they represent the same molecule, so compare canonical forms)
original_canonical = Chem.CanonSmiles(original_smiles)
predicted_canonical = Chem.CanonSmiles(predicted_smiles)
print("Match:", original_canonical == predicted_canonical)
```

```python
def validate_DECIMER_pipeline(smile_string: str):
    if isinstance(smile_string, str):
        smile_string = [smile_string]

    results = []
    for smi in smile_string:
        # Generate image from SMILES string
        mol = Chem.MolFromSmiles(smi)
        temp_img_path = "temp_molecule.png"
        Draw.MolToFile(mol, temp_img_path, size = (400,400), wedgeBonds=True)

        # Run DECIMER on generated image
        predicted_SMILEstring = predict_SMILES(temp_img_path)

        # Validate original string and Decimer's guess
        original_validation = validate_molecule(smi)
        predicted_validation = validate_molecule(predicted_SMILEstring)
        canonical_match = Chem.CanonSmiles(smi) == Chem.CanonSmiles(predicted_SMILEstring)
        
        # Output
        results.append({
            "Original SMILE string": smi,      
            "Predicted SMILE string": predicted_SMILEstring,
            "Canonical match": canonical_match,
            "Original validation": original_validation,
            "Predicted validation": predicted_validation
        })

    return results
```

## Function: Grading Stereocenter answer

```python
def grade_stereocenter_answer(correct_smiles: str, atom_index: int, student_answer: str):
    if not isinstance(student_answer, str):
        return {"error": f"Answer must be a string, got: {type(student_answer).__name__}"}

    cleaned_answer = student_answer.strip().upper()

    result = validate_molecule(correct_smiles)
    stereocenters = result[0]["Stereocenters"]

    correct_label = None
    for center in stereocenters:
        if center["Atom index"] == atom_index:
            correct_label = center["label"]

    if correct_label is None:
        return {"error": f"No stereocenter found at atom index {atom_index}"}

    is_correct = (cleaned_answer == correct_label)
    return {
        "correct_label": correct_label,
        "student_answer": cleaned_answer,
        "is_correct": is_correct
    }
```

```python
print(grade_stereocenter_answer("C[C@H](O)C(=O)O", 1, "S"))    # correct, clean input
print(grade_stereocenter_answer("C[C@H](O)C(=O)O", 1, "r"))    # wrong, lowercase input
print(grade_stereocenter_answer("C[C@H](O)C(=O)O", 1, " s "))  # correct, messy whitespace
print(grade_stereocenter_answer("C[C@H](O)C(=O)O", 5, "S"))    # bad atom index, should error cleanly
```

# **New Validation Function**
## Also finds "potential" stereocenters, not just confirmed ones

```python
def validate_molecule(smile_string: str):
    if isinstance(smile_string, str):
        smile_string = [smile_string]
    
    results = []

    for smi in smile_string:
        if not smi or not smi.strip():
            results.append({
                "Valid": False,
                "Number of stereocenters": 0,
                "Stereocenters": [],
                "Undefined stereocenters": []
            })
            continue 

        mol = Chem.MolFromSmiles(smi)
        if mol is None:
            results.append({
                "Valid":False, 
                "Number of stereocenters": 0,
                "Stereocenters": [],
                "Undefined stereocenters": []
                })
            continue

        # converts molecule to RDKit canon SMILES string and re-parse canon string back to molecule so atoms are numbered in canonical order
        canon_smiles = Chem.MolToSmiles(mol)
        mol = Chem.MolFromSmiles(canon_smiles)

        # add explicit Hs so we can see all 4 real substituents on a stereocenter
        # (not just the heavy atoms) -- needed for CIP priority ranking below.
        # AddHs keeps existing heavy-atom indices the same, only appends new H indices.
        mol = Chem.AddHs(mol)
        Chem.AssignStereochemistry(mol, cleanIt=True, force=True)
        rdCIPLabeler.AssignCIPLabels(mol)

        stereocenters = []
        for atom in mol.GetAtoms():
            if atom.HasProp("_CIPCode"):
                center = {
                    "Atom index": atom.GetIdx(),
                    "label": atom.GetProp("_CIPCode")
                }

                # CIP priority ranking of the four groups on this stereocenter,
                # highest to lowest, straight from RDKit's own CIP computation.
                if atom.HasProp("_CIPNeighborOrder"):
                    order = list(atom.GetPropsAsDict()["_CIPNeighborOrder"])
                    priorities = []
                    for rank, nbr_idx in enumerate(order, start=1):
                        nbr = mol.GetAtomWithIdx(nbr_idx)
                        attached = [n.GetSymbol() for n in nbr.GetNeighbors() if n.GetIdx() != atom.GetIdx()]
                        priorities.append({
                            "priority_rank": rank,  # 1 = highest priority
                            "atom_symbol": nbr.GetSymbol(),
                            "attached_to": attached,
                        })
                    center["Priorities"] = priorities

                stereocenters.append(center)

        all_centers = FindMolChiralCenters(mol, includeUnassigned=True, useLegacyImplementation= False)
        undefined_stereocenters = [idx for idx,label in all_centers if label == "?"]

        results.append({
                "Valid": True, 
                "Number of stereocenters": len(stereocenters), 
                "Stereocenters": stereocenters,
                "Undefined stereocenters": undefined_stereocenters
        })
    return results
```

```python
print(validate_molecule("CC(O)C(=O)O"))          # undefined stereochemistry
print(validate_molecule("CCO"))                   # genuinely no stereocenter at all
print(validate_molecule("C[C@H](O)C(=O)O"))       # fully defined
```

```python
test_cases = [
    ("Valid, no stereocenters", "CCO",
        {"Valid": True, "Number of stereocenters": 0}),
    ("Valid, one stereocenter", "C[C@H](O)C(=O)O",
        {"Valid": True, "Number of stereocenters": 1}),
    ("Valid, two stereocenters", "C[C@H](Cl)[C@@H](Br)C",
        {"Valid": True, "Number of stereocenters": 2}),
    ("Invalid SMILES (garbage)", "garbage!!!",
        {"Valid": False, "Number of stereocenters": 0}),
    ("Empty string", "",
        {"Valid": False, "Number of stereocenters": 0}),
    ("Undefined stereochemistry (no wedge/dash)", "CC(O)C(=O)O",
        {"Valid": True, "Number of stereocenters": 0}),
    ("Fake stereocenter (duplicate substituents)", "C[C@H](Br)C",
        {"Valid": True, "Number of stereocenters": 0}),
    ("Salt / multi-component SMILES", "[Na+].C[C@H](O)C(=O)[O-]",
        {"Valid": True, "Number of stereocenters": 1}),
    ("Partially-defined tartaric acid", "OC(=O)[C@H](O)C(O)C(=O)O",
        {"Valid": True, "Number of stereocenters": 1}),
    ("Combined R/S stereocenter + E/Z double bond", "C[C@H](O)/C=C/C",
        {"Valid": True, "Number of stereocenters": 1}),
]

correct = 0
total = len(test_cases)

for name, smi, expected in test_cases:
    result = validate_molecule(smi)[0]
    passed = (result["Valid"] == expected["Valid"] and
              result["Number of stereocenters"] == expected["Number of stereocenters"])
    correct += passed
    status = "PASS" if passed else "FAIL"
    print(f"[{status}] {name}")
    print(f"  input: {smi!r}")
    print(f"  result: {result}")
    print()

accuracy = (correct / total) * 100
print(f"Accuracy: {correct}/{total} correct ({accuracy:.1f}%)")
```

### Computed context for prompt

```python
def get_substituent_priorities(smiles: str, atom_index: int) -> list[dict]:
    result = validate_molecule(smiles)[0]
    for center in result["Stereocenters"]:
        if center["Atom index"] == atom_index:
            return center.get("Priorities", [])
    return []

def get_atomic_number(symbol: str) -> int:
    return Chem.GetPeriodicTable().GetAtomicNumber(symbol)

def build_priority_reason(priorities: list[dict]) -> str:
    """Deterministic, computed reason -- not something the LLM decides."""
    if len(priorities) < 2:
        return ""

    top, second = priorities[0], priorities[1]
    top_num = get_atomic_number(top["atom_symbol"])
    second_num = get_atomic_number(second["atom_symbol"])

    if top_num != second_num:
        return (
            f"{top['atom_symbol']} (atomic number {top_num}) outranks "
            f"{second['atom_symbol']} (atomic number {second_num}) because it has "
            f"the higher atomic number."
        )

    # Same atom, tied -- CIP breaks the tie using what each is attached to next.
    def best_attached_number(p):
        nums = [get_atomic_number(a) for a in p["attached_to"]]
        return max(nums) if nums else 0

    top_best = best_attached_number(top)
    second_best = best_attached_number(second)
    return (
        f"{top['atom_symbol']} and {second['atom_symbol']} tie on atomic number, "
        f"so the tie is broken by what they're attached to: the group attached to "
        f"{top['attached_to']} (highest atomic number {top_best}) outranks the one "
        f"attached to {second['attached_to']} (highest atomic number {second_best})."
    )

# R/S is *defined* as this rotation direction -- not a separate computation.
# Orient the molecule with the lowest-priority group pointing away from you,
# then trace priority 1 -> 2 -> 3: clockwise is R, counterclockwise is S.
ROTATION_BY_LABEL = {"R": "clockwise", "S": "counterclockwise"}

def describe_rotation(label: str) -> str:
    """Deterministic lookup, not something the LLM decides."""
    return ROTATION_BY_LABEL[label]
```

### LLM Grading Explanation Response

```python
def build_explanation_prompt(grading_result: dict, smiles: str, priorities: list[dict] | None = None, rotation: str | None = None) -> str:
    priority_block = ""
    if priorities:
        lines = "\n".join(
            f"  {p['priority_rank']}. {p['atom_symbol']}"
            f" (attached to: {', '.join(p['attached_to']) if p['attached_to'] else 'nothing else'})"
            for p in priorities
        )
        reason = build_priority_reason(priorities)
        priority_block = (
            "\n- Substituent priority order at this stereocenter, highest to lowest "
            f"(from CIP rules):\n{lines}"
            f"\n- Why the top group outranks the next one: {reason}"
        )

    rotation_block = ""
    if rotation:
        rotation_block = (
            "\n- Geometric fact (from CIP rules): with the lowest-priority group "
            f"pointing away from you, tracing priority 1 -> 2 -> 3 goes {rotation}, "
            f"which is what defines this as {grading_result['correct_label']}."
        )

    return f"""### Role
You are a friendly, encouraging organic chemistry tutor helping a student learn stereochemistry.

### Context
- Molecule (SMILES): {smiles}
- Student's answer: {grading_result['student_answer']}
- Correct answer: {grading_result['correct_label']}
- Was the student correct?: {grading_result['is_correct']}{priority_block}{rotation_block}

### Task
Write a short explanation (3-4 sentences) telling the student whether they got it right or wrong.
If a priority order, reason, or geometric fact are listed above, restate them in your own words --
do not compute or invent a new one.

### Rules (important)
- Only use the facts listed in Context above. Do NOT invent, guess, or compute any new chemistry
  beyond what's listed.
- Only mention a rotation direction (clockwise/counterclockwise) if a Geometric fact is given above,
  and use that exact direction -- never guess one, and never describe any other spatial detail
  (left/right, top/bottom, "points toward", wedge/dash) that isn't given to you word-for-word.
- If correct: briefly confirm it and give quick encouragement.
- If incorrect: clearly state the correct answer \u2014 never just say "wrong" without including it.
- Keep the tone encouraging, never harsh.

### Output format
Return ONLY the explanation text. No headers, no bullet points, no restating these instructions.
"""
```

```python
import requests 

def qwen_JSONresponse(prompt: str) -> dict:
    r = requests.post(
        "http://localhost:11434/api/generate",
        json={"model": "qwen3:1.7b", "prompt": prompt, "stream": False},
        timeout=60
    )
    data = r.json()
    return {
        "text": data["response"],
        "thinking": data.get("thinking", ""),
        "truncated": data.get("done_reason") != "stop",
        "duration_sec": data.get("eval_duration", 0) / 1e9,
    }
```

### Check LLM response

```python
import re

def check_explanation(grading_result: dict, response: dict, priorities: list[dict] | None = None, rotation: str | None = None) -> dict:
    text = response["text"]
    checks = {}

    checks["non_empty"] = bool(text.strip())
    checks["not_truncated"] = not response.get("truncated", False)

    label_pattern = re.compile(rf"(?<![A-Za-z]){re.escape(grading_result['correct_label'])}(?![A-Za-z])")
    checks["states_correct_label"] = bool(label_pattern.search(text))

    if not grading_result["is_correct"]:
        wrong_answer_as_correct = re.compile(
            rf"correct answer is\s*\**{re.escape(grading_result['student_answer'])}\**",
            re.IGNORECASE
        )
        checks["does_not_affirm_wrong_answer"] = not bool(wrong_answer_as_correct.search(text))

    # Position/layout claims (left/right, top/bottom, wedge/dash, etc.) are
    # never granted in any form, so they're always fabrication. "top"/
    # "bottom"/"right" are phrase-gated since the bare words also mean
    # "highest-ranked"/"correct", not just spatial.
    #
    # Words like "rotation", "orientation", "points away" are deliberately
    # NOT checked here -- they're descriptive filler, not independently
    # checkable claims (there's no "wrong" version of the word "rotation").
    # The only actual checkable claim is the specific direction named, which
    # is verified directly below against `rotation`.
    always_blocked_pattern = re.compile(
        r"\b(left|"
        r"on (the )?top\b|at the top\b|top side|top-hand|"
        r"on (the )?bottom\b|at the bottom\b|bottom side|"
        r"in front of|at the front|front side|"
        r"behind|above|below|"
        r"on the right|to the right|right side|right-hand|"
        r"points?\s+toward|pointing\s+toward|faces?\s+(toward|away)|"
        r"toward(s)?|closer to|farther (from|than)|nearest|farthest|"
        r"on the (wedge|dash)|spatial arrangement|3d arrangement)\b",
        re.IGNORECASE,
    )
    directional_ok = not bool(always_blocked_pattern.search(text))

    rotation_words = {
        m.lower().replace("-", "")
        for m in re.findall(r"\b(?:counter-?clockwise|clockwise)\b", text, re.IGNORECASE)
    }

    if len(rotation_words) > 1:
        # claimed both directions for one stereocenter -- contradictory, never valid
        rotation_ok = False
    elif rotation is None:
        rotation_ok = not rotation_words
    else:
        wrong_word = "clockwise" if rotation == "counterclockwise" else "counterclockwise"
        rotation_ok = wrong_word not in rotation_words

    checks["no_spatial_fabrication"] = rotation_ok and directional_ok

    if priorities is None:
        # No grounded substituent data was given this time -- any chemistry-reasoning
        # vocabulary at all means it invented the explanation from nothing.
        fabrication_flags = [
            "substituent", "priority", "outrank", "identical",
            "because the", "chiral center is", "double bond", "hydroxyl",
            "highest priority", "lowest priority", "attached group", "functional group",
            "electronegat",
        ]
        checks["no_fabricated_reasoning"] = not any(flag in text.lower() for flag in fabrication_flags)

    checks["passed"] = all(v for k, v in checks.items() if k != "passed")
    return checks
```

```python
def get_safe_explanation(grading_result: dict, smiles: str, atom_index: int | None = None) -> dict:
    if "error" in grading_result:
        return {
            "text": f"Couldn't grade that: {grading_result['error']}",
            "used_fallback": True,
            "raw_response": None,
            "checks": None,
        }

    label = grading_result["correct_label"]
    fallback_text = (
        f"Correct! The stereocenter is {label}."
        if grading_result["is_correct"]
        else f"Not quite \u2014 the correct answer is {label}."
    )

    priorities = get_substituent_priorities(smiles, atom_index) if atom_index is not None else None
    priorities = priorities or None  # treat an empty list the same as "no data available"
    rotation = describe_rotation(label) if priorities else None  # only ground rotation when there's a priority order to hang it on

    # Rebuild the prompt
    prompt = build_explanation_prompt(grading_result, smiles, priorities, rotation)

    try:
        response = qwen_JSONresponse(prompt)
    except requests.exceptions.RequestException as e:
        # Ollama down, slow, or unreachable -- same fallback as a failed content check,
        # since a crash is worse than a plain-but-correct answer.
        return {
            "text": fallback_text,
            "used_fallback": True,
            "raw_response": None,
            "checks": None,
            "error": str(e),
        }

    checks = check_explanation(grading_result, response, priorities, rotation)

    if checks["passed"]:
        return {
            "text": response["text"],
            "used_fallback": False,
            "raw_response": response,
            "checks": checks,
            "priorities": priorities,
            "rotation": rotation,
        }

    return {
        "text": fallback_text,
        "used_fallback": True,
        "raw_response": response,
        "checks": checks,
        "priorities": priorities,
        "rotation": rotation,
    }
```

```python
llm_test_cases = [
    ("Wrong answer (the exact case that hallucinated earlier)", "C[C@H](O)C(=O)O", 1, "R"),
    ("Correct answer", "C[C@H](O)C(=O)O", 1, "S"),
    ("Bad atom index (error case)", "C[C@H](O)C(=O)O", 5, "S"),
]

for name, smi, idx, ans in llm_test_cases:
    grading = grade_stereocenter_answer(smi, idx, ans)
    result = get_safe_explanation(grading, smi, atom_index=idx)
    print(f"[{name}]")
    print(f"  grading: {grading}")
    print(f"  used_fallback: {result['used_fallback']}")
    if result["checks"] is not None:
        print(f"  checks: {result['checks']}")
    print(f"  text: {result['text']}")
    print()
```

### Expanded LLM Explanation Test Set
Covers gaps the 3-case smoke test didn't: error paths that must never reach the LLM, multi-stereocenter scoping, an atomic-number tie-break, and charged/salt SMILES. Fischer projections and hand-drawn input are deliberately out of scope for this pass.

```python
llm_test_cases_v2 = [
    # -- regression cases from the original smoke test, kept for continuity --
    ("Wrong answer, single stereocenter", "C[C@H](O)C(=O)O", 1, "R"),
    ("Correct answer, single stereocenter", "C[C@H](O)C(=O)O", 1, "S"),
    ("Bad atom index (error path, no LLM call)", "C[C@H](O)C(=O)O", 5, "S"),

    # -- error paths that must short-circuit before ever calling the LLM --
    ("Achiral molecule, no stereocenter at all", "CCO", 1, "S"),
    ("Undefined stereochemistry (no wedge/dash drawn)", "CC(O)C(=O)O", 1, "S"),

    # -- multi-stereocenter: does the explanation stay scoped to ONE center? --
    ("Two-stereocenter molecule (threonine), correct answer for center 1", "C[C@H]([C@@H](C(=O)O)N)O", 1, "R"),
    ("Two-stereocenter molecule (threonine), wrong answer for center 2", "C[C@H]([C@@H](C(=O)O)N)O", 3, "R"),
    ("Three-stereocenter molecule (menthol), correct answer for one center", "CC(C)[C@@H]1CC[C@@H](C)C[C@H]1O", 3, "S"),

    # -- atomic-number TIE at the top two priorities -- exercises the
    # tie-break branch of build_priority_reason, never hit by the original set --
    ("Tie-break case: all-carbon substituents (3-methylhexane)", "CC[C@@H](C)CCC", 3, "R"),

    # -- charged / multi-fragment edge cases --
    ("Charged zwitterion stereocenter (alanine)", "C[C@@H]([NH3+])C(=O)[O-]", 1, "R"),
    ("Salt / spectator ion in the SMILES", "[Na+].C[C@H](O)C(=O)[O-]", 1, "R"),  # wrong answer, real is S
]

reached_llm = 0
passed = 0
for name, smi, idx, ans in llm_test_cases_v2:
    grading = grade_stereocenter_answer(smi, idx, ans)
    result = get_safe_explanation(grading, smi, atom_index=idx)
    print(f"[{name}]")
    print(f"  grading: {grading}")
    print(f"  used_fallback: {result['used_fallback']}")
    if result.get("checks") is not None:
        reached_llm += 1
        passed += result["checks"]["passed"]
        print(f"  checks: {result['checks']}")
    print(f"  text: {result['text']}")
    print()

print(
    f"LLM explanation accepted on {passed}/{reached_llm} cases that reached the LLM "
    f"({len(llm_test_cases_v2) - reached_llm} cases correctly short-circuited before calling it)."
)
```
