### General function for testing

```python
def validate_molecule(smile_string: str):
    if isinstance(smile_string, str):
        smile_string = [smile_string]

    results = []
    for smi in smile_string:
        mol = Chem.MolFromSmiles(smi)
        if mol is None:
            results.append({"Valid":False, "Stereocenters": []})
            continue

        rdCIPLabeler.AssignCIPLabels(mol)

        stereocenters = []
        for atom in mol.GetAtoms():
            if atom.HasProp("_CIPCode"):
                stereocenters.append({
                    "Atom index": atom.GetIdx(),
                    "label": atom.GetProp("_CIPCode")
                })
        results.append({"Valid": True, 
               "Number of stereocenters": len(stereocenters), 
               "Stereocenters": stereocenters })

    return results
```

### Decimer Pipeline function

```python
from rdkit import Chem
from rdkit.Chem import Draw
from DECIMER import predict_SMILES

# Step 1: generate a test image of a known molecule (aspirin)
original_smiles = "CC(=O)Oc1ccccc1C(=O)O"
mol = Chem.MolFromSmiles(original_smiles)
Draw.MolToFile(mol, "aspirin_test.png", size=(400, 400))

# Step 2: feed it through DECIMER
predicted_smiles = predict_SMILES("aspirin_test.png")
print("Original: ", original_smiles)
print("Predicted:", predicted_smiles)

# Step 3: compare using RDKit (SMILES strings can differ in formatting
# even when they represent the same molecule, so compare canonical forms)
original_canonical = Chem.CanonSmiles(original_smiles)
predicted_canonical = Chem.CanonSmiles(predicted_smiles)
print("Match:", original_canonical == predicted_canonical)
```

```python
def validate_DECIMER_pipeline(smile_string: str):
    if isinstance(smile_string, str):
        smile_string = [smile_string]

    results = []
    for smi in smile_string:
        # Generate image from SMILES string
        mol = Chem.MolFromSmiles(smi)
        temp_img_path = "temp_molecule.png"
        Draw.MolToFile(mol, temp_img_path, size = (400,400), wedgeBonds=True)

        # Run DECIMER on generated image
        predicted_SMILEstring = predict_SMILES(temp_img_path)

        # Validate original string and Decimer's guess
        original_validation = validate_molecule(smi)
        predicted_validation = validate_molecule(predicted_SMILEstring)
        canonical_match = Chem.CanonSmiles(smi) == Chem.CanonSmiles(predicted_SMILEstring)
        
        # Output
        results.append({
            "Original SMILE string": smi,      
            "Predicted SMILE string": predicted_SMILEstring,
            "Canonical match": canonical_match,
            "Original validation": original_validation,
            "Predicted validation": predicted_validation
        })

    return results
```

```python
result = validate_DECIMER_pipeline("C[C@H](O)C(=O)O")
print(result)
```

```python
test_molecules = [
    "CCO",                                     # Ethanol
    "CC(=O)C",                                 # Acetone
    "c1ccccc1",                                # Benzene
    "CC(=O)Oc1ccccc1C(=O)O",                   # Aspirin
    "C[C@H](O)C(=O)O",                         # (S)-Lactic acid
    "C[C@@H](O)C(=O)O",                        # (R)-Lactic acid
    "C[C@@H](C(=O)O)N",                        # (S)-Alanine
    "CC[C@@H](C)O",                            # (R)-2-Butanol
    "CC[C@@H](Cl)C",                           # (S)-2-Chlorobutane
    "[C@](F)(Cl)(Br)I",                        # Bromochlorofluoroiodomethane
    "OC[C@@H](O)C=O",                          # (R)-Glyceraldehyde
    "C([C@@H](C(=O)O)N)O",                     # (S)-Serine
    "CC(C)Cc1ccc(cc1)[C@H](C)C(=O)O",          # (S)-Ibuprofen
    "OC(=O)[C@H](O)[C@@H](O)C(=O)O",           # (2R,3R)-Tartaric acid
    "OC(=O)[C@H](O)[C@H](O)C(=O)O",            # meso-Tartaric acid
    "C[C@H]([C@@H](C(=O)O)N)O",                # Threonine
    "CC(C)[C@@H]1CC[C@@H](C)C[C@H]1O",         # Menthol
    "CC[C@H](N)C",                             # sec-Butylamine
    "C/C=C/C",                                 # (E)-2-Butene
    "C/C=C\\C",                                # (Z)-2-Butene
]
```

```python
results = validate_DECIMER_pipeline(test_molecules)

for smi, result in zip(test_molecules, results):
    print(f"{smi} -> {result}\n")
```
