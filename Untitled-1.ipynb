```python
from rdkit import Chem
from rdkit.Chem import Draw
from DECIMER import predict_SMILES

# Step 1: generate a test image of a known molecule (aspirin)
original_smiles = "CC(=O)Oc1ccccc1C(=O)O"
mol = Chem.MolFromSmiles(original_smiles)
Draw.MolToFile(mol, "aspirin_test.png", size=(400, 400))

# Step 2: feed it through DECIMER
predicted_smiles = predict_SMILES("aspirin_test.png")
print("Original: ", original_smiles)
print("Predicted:", predicted_smiles)

# Step 3: compare using RDKit (SMILES strings can differ in formatting
# even when they represent the same molecule, so compare canonical forms)
original_canonical = Chem.CanonSmiles(original_smiles)
predicted_canonical = Chem.CanonSmiles(predicted_smiles)
print("Match:", original_canonical == predicted_canonical)
```

### General function for testing

```python
def validate_molecule(smile_string: str):
    if isinstance(smile_string, str):
        smile_string = [smile_string]

    results = []
    for smi in smile_string:
        mol = Chem.MolFromSmiles(smi)
        if mol is None:
            results.append({"Valid":False, "Stereocenters": []})
            continue

        rdCIPLabeler.AssignCIPLabels(mol)

        stereocenters = []
        for atom in mol.GetAtoms():
            if atom.HasProp("_CIPCode"):
                stereocenters.append({
                    "Atom index": atom.GetIdx(),
                    "label": atom.GetProp("_CIPCode")
                })
        results.append({"Valid": True, 
               "Number of stereocenters": len(stereocenters), 
               "Stereocenters": stereocenters })

    return results
```

Test 1: One stereocenter

```python
from rdkit.Chem import rdCIPLabeler

for smi in ["C[C@@H](O)C(=O)O", "C[C@H](O)C(=O)O"]:
    mol = Chem.MolFromSmiles(smi)
    rdCIPLabeler.AssignCIPLabels(mol)
    for atom in mol.GetAtoms():
        if atom.HasProp("_CIPCode"):
            print(smi, "-->", atom.GetProp("_CIPCode"))
```

```python
# Alternative function
validate_molecule(["C[C@@H](O)C(=O)O", "C[C@H](O)C(=O)O"])
```

Test 2: Two stereocenters

```python
smi = "C[C@H](Cl)[C@@H](Br)C"
mol_two_sterocenters = Chem.MolFromSmiles(smi)
rdCIPLabeler.AssignCIPLabels(mol_two_sterocenters)

for atom in mol_two_sterocenters.GetAtoms():
    if atom.HasProp("_CIPCode"):
        print(f"Atom index {atom.GetIdx()} -> {atom.GetProp("_CIPCode")}")
```

```python
validate_molecule(["C[C@H](Cl)[C@@H](Br)C"])
```
